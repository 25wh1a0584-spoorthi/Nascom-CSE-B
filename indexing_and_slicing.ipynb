```python
Define a list
my_list = [10, 20, 30, 40, 50]#
```

```python
# Indexing (0-based)
print("First element:", my_list[0])   # 10
print("Last element:", my_list[-1])   # 50
```

```python
# Define a tuple
my_tuple = ("a", "b", "c", "d", "e")
```

```python
# Indexing
print("Element at index 0:", my_tuple[0])
print("Element at index -2:", my_tuple[-2])  # 'd'

```
